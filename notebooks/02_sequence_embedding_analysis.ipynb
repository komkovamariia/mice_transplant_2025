# Approach 2: sequence-space enrichment

A shared TCREmp/PCA basis is used for local-density enrichment, RFF-MMD, PERMANOVA and witness scoring. Results are computed separately for each biological stratum.

```python
from src.runtime import configure_runtime

N_WORKERS = configure_runtime()

from src.approach2_sequence_embedding import compile_summary, prepare_embedding, run_stratum
from src.figures import write_figure_inventory
from src.strata import load_repertoire, requested_strata, validate_strata

repertoire = load_repertoire()
STRATA = requested_strata()
print(f"Workers available: {N_WORKERS}")
display(validate_strata(repertoire, STRATA))
```

```python
context = prepare_embedding(repertoire, seed=0)
print(
    f"Shared sequence-space basis: {context['coordinates'].shape[0]:,} "
    f"clonotypes x {context['coordinates'].shape[1]} dimensions"
)
```
